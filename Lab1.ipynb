Задача 1

```python
from random import randint
n=randint(10,21)
mas1=tuple(randint(10,100) for i in range(n))
for i in mas1:
    print(i,end=" ")
print()
summa=0
for i in mas1:
    summa+=i
print("s=",summa,sep="")
```

Задача 2


```python
n,m=5,6
matrix=tuple(tuple(randint(10,100) for j in range(m)) for i in range(n))
for i in matrix:
    print(*i)
summa_d=0
for i in range(len(matrix)):
    for j in range(len(matrix[i])):
        if i==j:
            summa_d+=matrix[i][j]
print("s=",summa_d,sep="")
```

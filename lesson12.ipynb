```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
```

```python
texts = ["I love this product", "This is the worst thing ever", "Absolutely amazing!", "Terrible,do not buy"]
labels = ["positive", "negative", "positive", "negative"]
```

```python
vectorizer = TfidfVectorizer()
```

```python
X = vectorizer.fit_transform(texts)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, labels, test_size=0.25)
```

```python
X_test
```

```python
knn = KNeighborsClassifier(n_neighbors=3)
```

```python
knn.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)
y_pred
```

```python
print("Predictions:", y_pred)
```

Процесс работает так:
1. Каждому объекту (например, фильму) присваиваются признаки, такие как жанр, год
выпуска, актёры и т. д.
2. Когда пользователь оценивает фильм, система находит фильмы с похожими признаками и рекомендует их пользователю.

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np
```

```python
movies = np.array([
[1, 0, 1], # Фильм 1: жанр 1, жанр 3
[1, 1, 0], # Фильм 2: жанр 1, жанр 2
[0, 1, 1], # Фильм 3: жанр 2, жанр 3
[1, 0, 0], # Фильм 4: жанр 1
])
```

```python
knn = NearestNeighbors(n_neighbors=2)
```

```python
knn.fit(movies)
```

```python
distances, indices = knn.kneighbors([[1, 0, 1]])
```

```python
print("Nearest Movies:", indices)
```

Процесс включает следующие шаги:
1. Для каждого дома создаются признаки, такие как площадь, количество комнат, возраст
здания и т. д.
2. Когда необходимо предсказать цену для нового дома, алгоритм находит K ближайших
домов, для которых известна цена, и выдает среднее значение цен этих домов как предсказание.

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import train_test_split
```

```python
X = np.array([[1000, 2], [1500, 3], [2000, 4], [2500, 5], [3000, 6]])
y = np.array([200000, 300000, 400000, 500000, 600000])
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25)
```

```python
knn_regressor = KNeighborsRegressor(n_neighbors=2)
```

```python
knn_regressor.fit(X_train, y_train)
```

```python
price_pred = knn_regressor.predict([[1200, 3]])
```

```python
print("Predicted Price:", price_pred)
```

Классификация рукописных цифр с
использованием набора данных MNIST.

```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
```

```python
digits=datasets.load_digits()
digits
```

```python
X=digits.data
y=digits.target
```

```python
X
```

```python
y
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
scaler=StandardScaler()
```

```python
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
knn=KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train_scaled, y_train)
```

```python
y_pred=knn.predict(X_test_scaled)
```

```python
accuracy=accuracy_score(y_test,y_pred)
```

```python
print(f"Accuracy: {accuracy * 100:.2f}%")
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score
```

```python
data = load_iris()
X = data.data # Признаки (длина и ширина лепестков и чашелистиков)
y = data.target # Целевая переменная (виды цветов)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
knn = KNeighborsClassifier(n_neighbors=3)
```

```python
knn.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)
```

```python
# Оценка точности модели
accuracy = accuracy_score(y_test, y_pred)
print(f"Точность модели: {accuracy * 100:.2f}%")
```

```python
new_flower = np.array([[5.2, 2.0, 3.5, 1.0]])
```

```python
predicted_class = knn.predict(new_flower)
print(f"Предсказанный вид для нового цветка: {data.target_names[predicted_class][0]}")
```
